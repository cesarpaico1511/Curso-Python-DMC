# Ejercicio 1: Cálculo del Consumo de Alimentos

Este programa simula el proceso de facturación de un restaurante, calculando automáticamente una propina del 10% sobre el valor del consumo y generando un resumen detallado con la fecha actual del sistema.

```python
from datetime import datetime

# Solicitar información al usuario
nombre_mesero = input("Ingrese el nombre del mesero/a: ")
valor_consumo = float(input("Ingrese el valor del consumo: "))

# Definir el porcentaje de propina (10%)
propina_tasa = 0.10

# Realizar los cálculos necesarios
valor_propina = valor_consumo * propina_tasa
total_a_pagar = valor_consumo + valor_propina

# Obtener y formatear la fecha actual
fecha_actual = datetime.now().strftime("%d/%m/%Y")

# Imprimir resumen de la transacción usando f-strings
print("\n" + "*"*40)
print("CÁLCULO DEL CONSUMO DE ALIMENTOS")
print("*"*40)
print(f"Mesero/a: {nombre_mesero}")
print(f"Valor base: ${valor_consumo:.2f}")
print(f"Valor de la propina: ${valor_propina:.2f}")
print(f"Valor total a pagar: ${total_a_pagar:.2f}")
print(f"Fecha de consulta: {fecha_actual}")
print("*"*40)
```
